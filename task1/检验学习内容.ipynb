```python
# Jupyter Notebook 示例代码
# 可以直接在Jupyter环境中运行

# 导入常用库
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 设置中文显示
plt.rcParams["font.family"] = ["SimHei", "WenQuanYi Micro Hei", "Heiti TC"]
sns.set(font="SimHei", font_scale=1.2)

# 1. 生成随机数据并进行基本统计
print("===== 随机数据统计 =====")
# 生成100个符合正态分布的随机数
data = np.random.normal(loc=50, scale=10, size=100)
print(f"数据样本: {data[:5]}...")
print(f"平均值: {np.mean(data):.2f}")
print(f"标准差: {np.std(data):.2f}")
print(f"最大值: {np.max(data):.2f}")
print(f"最小值: {np.min(data):.2f}")

# 2. 创建DataFrame并进行简单分析
print("\n===== DataFrame示例 =====")
df = pd.DataFrame({
    '类别': ['A']*50 + ['B']*50,
    '数值': np.concatenate([
        np.random.normal(60, 15, 50),
        np.random.normal(40, 10, 50)
    ])
})
print(df.head())
print("\n按类别统计:")
print(df.groupby('类别')['数值'].describe())

# 3. 数据可视化
print("\n===== 数据可视化 =====")
plt.figure(figsize=(12, 5))

# 子图1: 直方图
plt.subplot(1, 2, 1)
sns.histplot(data=data, kde=True, color='skyblue')
plt.title('随机数据分布')
plt.xlabel('数值')
plt.ylabel('频数')

# 子图2: 箱线图
plt.subplot(1, 2, 2)
sns.boxplot(x='类别', y='数值', data=df)
plt.title('不同类别的数值分布')
plt.xlabel('类别')
plt.ylabel('数值')

plt.tight_layout()
plt.show()

# 4. 简单的交互示例
def calculate_area(radius):
    """计算圆的面积"""
    return np.pi * radius **2

print("\n===== 函数示例 =====")
radius = 5
print(f"半径为{radius}的圆面积是: {calculate_area(radius):.2f}")
```
